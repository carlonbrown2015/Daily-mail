```python
import smtplib
from email.mime.text import MIMEText
```

```python
def send_mail():
    sender = "carlonbrown2015@gmail.com"   # your Gmail
    recipient = "carlonbrown2015@gmail.com"
    app_password = "hpdx frut lxom lult"  # from Google App Passwords

    msg = MIMEText("it works you SOB, it works")
    msg["Subject"] = "Test from Jupyter Scheduler"
    msg["From"] = sender
    msg["To"] = recipient

    try:
        with smtplib.SMTP_SSL("smtp.gmail.com", 465) as server:
            server.login(sender, app_password)
            server.sendmail(sender, [recipient], msg.as_string())
        print("✅ Email sent successfully")
    except Exception as e:
        print("❌ Error:", e)

send_mail()
```
